# Ewald Summation on Quantum Computer

A hybrid quantum–classical implementation of the Ewald summation for the
electrostatic energy of a periodic 3D lattice of point charges. The algorithm employs the Ewald method based decomposition of electrostatic energy into several energy terms, of which the ''Fourier component'' (long-range electrostatics) computed on a quantum device, utilizing the power of Quantum Fourier Transform (QFT). 

The corresponding preprint is available on arXiv:
Mansur Ziiatdinov, Igor Novikov, Farid Ablayev, and Valeri Barsegov, Quantum Algorithm for Ewald Summation–Based Computation of Long-Range Electrostatics, arXiv. https://doi.org/10.48550/arXiv.2512.20886

The total energy is split as

$$ E \;=\; E^L \;+\; E^S \;+\; E^{\mathrm{self}} \;+\; E^{\mathrm{dipole}} $$

with

$$ E^L \;=\; \frac{1}{2\,V\,\varepsilon_0}
   \sum_{\mathbf{k}\neq 0}
   \frac{\exp(-\sigma^2 k^2 / 2)}{k^2}\, |\hat\rho(\mathbf{k})|^2, $$

$$ E^S \;=\; \frac{1}{4\pi\varepsilon_0}
   \sum_{\substack{i<j \\ \mathbf{n}}}
   \frac{q_i q_j}{r_{ij,\mathbf{n}}}\,
   \mathrm{erfc}\!\left(\frac{r_{ij,\mathbf{n}}}{\sigma\sqrt{2}}\right), $$

$$ E^{\mathrm{self}} \;=\; -\frac{1}{4\pi\varepsilon_0}\,
   \frac{1}{\sigma\sqrt{2\pi}}\sum_i q_i^2, $$

$$ E^{\mathrm{dipole}} \;=\; \frac{1}{4\pi\varepsilon_0}\,
   \frac{2\pi}{(1 + 2\varepsilon_p)\,V}\,|\mathbf{P}|^2, $$

 where $\sigma$ is the standard deviation of the Gaussian function, ${\bf k}$ is the wavevector ($k=\|{\bf k}\|$), $V$ is the volume of the supercell, and $\epsilon'$ is the dielectric constant of the medium. $\mathbf{P} = \sum_i q_i\,\mathbf{r}_i$ is the total dipole moment,
$\sigma$ is the Ewald split width, and $\mathbf{n}$ ranges over image
shells. The quantum step recovers $|\hat\rho(\mathbf{k})|^2$ as the
measurement-probability density of the state
$|\psi\rangle = \sum_{\mathbf{x}} \rho(\mathbf{x})\,|\mathbf{x}\rangle$
after a 3D QFT.
The sum over ${\bf n}$ takes into account the periodic images of the charges and the prime indicates that in case $i = j$ the term ${\bf n}=0$ must be omitted.

The calculation below is carried out on a cubic $M_x \times M_y \times M_z = 64 \times 64 \times 64$ grid, i.e. $6 + 6 + 6 = 18$ qubits, with a alternating-sign lattice of 30 752 point charges.

````python
import time
from collections import defaultdict

import numpy as np
from numpy.fft import fftn

from scipy.spatial.distance import cdist
from scipy.special import erfc

from qiskit import QuantumCircuit, transpile
from qiskit.circuit.library import QFT
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
from qiskit_aer.primitives import SamplerV2
````

## Convert input into state

The input is a set of point charges on an $M_x \times M_y \times M_z$
integer grid:

```
[(q_i, (x_i, y_i, z_i))] -> [alpha_j]
```

The cells below encode this distribution as a length-$2^{N_{\text{qubits}}}$
vector $\alpha_j$ whose entry at index $j = (x \,\Vert\, y \,\Vert\, z)_2$
holds the charge at position $(x, y, z)$. Concretely, the bitstring
encoding packs $x$ in the most-significant `MX_QUBITS` bits, then $y$ in
the next `MY_QUBITS`, then $z$ in the last `MZ_QUBITS`
(`M*_QUBITS = log_2 M_*`). The vector is then L2-normalized and loaded as
the amplitudes of an $N_{\text{qubits}}$-qubit state.

Charge $q_i \in [-W; W]$; coordinates $x_i \in [0; M_x)$,
$y_i \in [0; M_y)$, $z_i \in [0; M_z)$.

````python
# --- Simulation mode ---------------------------------------------------------
# True  -> ideal statevector simulation (no shot noise, no measurements added).
# False -> sampling-based simulation with N_SHOTS shots.
EXACT_SIM = False

# --- Physical constants (SI) -------------------------------------------------
eps0 = 8.854e-12          # Vacuum permittivity [F/m]
W    = 1.6e-19            # Charge magnitude (elementary charge) [C]

# --- Grid --------------------------------------------------------------------
MX   = 64                 # Grid points along x; each axis must be a power of 2
MY   = 64                 # Grid points along y
MZ   = 64                 # Grid points along z
D    = 3                  # Spatial dimensions
step = 0.75e-10           # Lattice spacing [m]  (same along all three axes)

# --- Ewald summation parameters ---------------------------------------------
SIGMA = 0.880e-10         # Real/reciprocal split width [m].  Scanned at 0.001 A resolution for
                          # this configuration: minimises rms_E against the direct Coulomb sum
                          # (3.42e-05) while keeping E^L a live term (~0.17% of the total energy).
RCUT  = 5.0e-10           # Real-space cutoff radius [m]
epsp  = 1                 # Surrounding-medium relative permittivity (dipole term)

# --- Periodic-image shell counts --------------------------------------------
n = 1                     # Image shells used in the real-space term E^S
m = 1                     # Image shells used in the direct Coulomb reference

# --- Quantum sampling -------------------------------------------------------
N_SHOTS = 10_000_000

# --- Chunking for the direct Coulomb reference ------------------------------
CHUNK = 100               # rows per block; controls peak memory, not the result
                          # (each block is a (CHUNK x 26N) distance matrix)
````

````python
MX_QUBITS = int(np.log2(MX))
MY_QUBITS = int(np.log2(MY))
MZ_QUBITS = int(np.log2(MZ))
N_QUBITS  = MX_QUBITS + MY_QUBITS + MZ_QUBITS      # 6 + 6 + 6 = 18

SHAPE  = (MX, MY, MZ)                 # grid shape of every real/reciprocal array below
N_GRID = MX * MY * MZ                 # total number of grid sites
BOX    = np.array([MX, MY, MZ]) * step        # box lengths per axis [m]
VOL    = MX * MY * MZ * step**3               # supercell volume [m^3]

# Qubit indices for each spatial coordinate register.
# The flat state index is (x << (MY_QUBITS + MZ_QUBITS)) | (y << MZ_QUBITS) | z,
# i.e. the bitstring 'x||y||z' with x in the most significant bits.  Qiskit puts
# bit q of that index on qubit q (qubit 0 = least significant), so z lives on the
# LOWEST qubit indices and x on the HIGHEST.  The registers are written this way
# round so that the assignment keeps following bit significance for any choice of
# M_x, M_y, M_z, not only for equal-size registers.
Z_REG = tuple(range(MZ_QUBITS))
Y_REG = tuple(q + MZ_QUBITS             for q in range(MY_QUBITS))
X_REG = tuple(q + MZ_QUBITS + MY_QUBITS for q in range(MX_QUBITS))

print(f"Grid {MX} x {MY} x {MZ} = {N_GRID:,} sites   qubits: {N_QUBITS}")
print(f"X_REG = {X_REG}   Y_REG = {Y_REG}   Z_REG = {Z_REG}")
print(f"Box = {BOX / 1e-10} A   V = {VOL:.4e} m^3")
````

````python
def pos_to_str(x: int, y: int, z: int) -> str:
    """Encode integer grid coordinates as a binary bitstring (one field per axis)."""
    return f'{x:0{MX_QUBITS}b}{y:0{MY_QUBITS}b}{z:0{MZ_QUBITS}b}'


def convert_input_to_pairs(charges):
    """Convert [(q, (x, y, z))] entries into [(q, bitstring)] entries."""
    return [(q, pos_to_str(*r)) for q, r in charges]
````

### Configuration

Build the charge list `inp` as a list of triples `(q, (x, y, z))`.
The configuration is a alternating-sign lattice placed as a
**centred box** — `place_box(32, 31, 31)`: $32 \times 31 \times 31 = 30\,752$
charges on the stride-2 sublattice (every second grid point), centred in the
$64 \times 64 \times 64$ box. 

Signs alternate as $(-1)^{i_x + i_y + i_z}$ over the occupied sublattice
indices. 

````python
def place_box(nx, ny, nz, centered=True):
    """A box of nx*ny*nz charges on the stride-2 sublattice (every 2nd grid point),
    centred in the M_x x M_y x M_z grid (corner-anchored if centered=False)."""
    XS, YS, ZS = range(0, MX, 2), range(0, MY, 2), range(0, MZ, 2)
    NXM, NYM, NZM = len(XS), len(YS), len(ZS)
    assert 1 <= nx <= NXM and 1 <= ny <= NYM and 1 <= nz <= NZM, \
        f"per-axis counts must lie in (1..{NXM}, 1..{NYM}, 1..{NZM})"
    def rng(nmax, n):
        s = (nmax - n) // 2 if centered else 0
        return range(s, s + n)
    return [(-W if (ai + aj + ak) % 2 == 0 else W, (XS[ai], YS[aj], ZS[ak]))
            for ai in rng(NXM, nx) for aj in rng(NYM, ny) for ak in rng(NZM, nz)]


inp = place_box(32, 31, 31)

print(inp[:4], '...', inp[-2:])
print(f"Number of charges: {len(inp)}")
xs = sorted({r[0] for _, r in inp}); ys = sorted({r[1] for _, r in inp}); zs = sorted({r[2] for _, r in inp})
print(f"Box extent:        x {xs[0]}..{xs[-1]} ({len(xs)}),  "
      f"y {ys[0]}..{ys[-1]} ({len(ys)}),  z {zs[0]}..{zs[-1]} ({len(zs)})")
print(f"Net charge:        {sum(q for q, _ in inp):.3e} C")
print(convert_input_to_pairs(inp)[:4])
````

### Visualize the charge configuration

Before running anything quantum, look at the system you just built. The two
helpers below plot the charge list `inp` directly (they do not touch any of the
energy code):

* **`plot_charges(inp)`** — a 3D scatter of the whole box. Red points are $+W$
  charges, blue are $-W$; the grey wireframe is the periodic
  $M_x \times M_y \times M_z$ supercell, drawn to scale. With 30 752 charges the
  scatter is dense, so pass `subsample=k` to plot a random $1/k$ of them
  (readability only — the box outline is always the full cell), at the cost of
  scrambling the visible +/- alternation.
* **`plot_slice(inp, axis='z', index=0)`** — one lattice plane.

Set `unit='ang'` on either to label the axes in ångström (grid index × `step`)
instead of integer grid points.


````python
import matplotlib.pyplot as plt
from itertools import product
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401  (registers the 3d projection)


def plot_charges(charges, subsample=None, ax=None, unit='grid'):
    """3D scatter of the charge configuration, coloured by the sign of the charge.

    Red = positive (+W), blue = negative (-W).  `subsample` (an int k) plots a
    random 1/k of the charges -- readability only, the box wireframe always shows
    the full MX x MY x MZ supercell.  `unit` is 'grid' (integer grid points) or
    'ang' (angstrom, = grid index * step).
    """
    q = np.array([c[0] for c in charges], dtype=float)
    xyz = np.array([c[1] for c in charges], dtype=float)
    scale = 1.0 if unit == 'grid' else step / 1e-10
    box = np.array([MX, MY, MZ]) * scale
    xyz = xyz * scale

    if subsample and subsample > 1:
        # Random (not strided) thinning, so the sample spans the whole box
        # rather than collapsing onto a few lattice planes.
        rng = np.random.default_rng(0)
        keep = rng.choice(len(q), size=max(1, len(q) // subsample), replace=False)
        xyz, q = xyz[keep], q[keep]

    if ax is None:
        fig = plt.figure(figsize=(8, 7))
        ax = fig.add_subplot(111, projection='3d')

    pos, neg = q > 0, q < 0
    ax.scatter(*xyz[pos].T, c='#d62728', s=8, alpha=0.6, label='+W')
    ax.scatter(*xyz[neg].T, c='#1f77b4', s=8, alpha=0.6, label='-W')

    # Box wireframe (the periodic supercell): connect corners that differ in one axis.
    corners = np.array(list(product([0, box[0]], [0, box[1]], [0, box[2]])))
    for a, b in product(range(8), range(8)):
        if a < b and np.count_nonzero(corners[a] - corners[b]) == 1:
            ax.plot(*zip(corners[a], corners[b]), c='0.4', lw=0.8)

    label = 'grid points' if unit == 'grid' else 'angstrom'
    ax.set_xlabel(f'x ({label})'); ax.set_ylabel(f'y ({label})'); ax.set_zlabel(f'z ({label})')
    ax.set_box_aspect(box)                      # draw the 64:32:32 ratio to scale
    shown = '' if len(q) == len(charges) else f'  ({len(q):,} shown)'
    ax.set_title(f'{MX} x {MY} x {MZ} box, {len(charges):,} charges{shown}')
    ax.legend(loc='upper left')
    return ax


def plot_slice(charges, axis='z', index=0, unit='grid', ax=None):
    """2D scatter of a single lattice plane (all charges with coordinate[axis] == index).

    The checkerboard +/- alternation of the NaCl-like lattice is clearest here.
    """
    ai = {'x': 0, 'y': 1, 'z': 2}[axis]
    other = [a for a in (0, 1, 2) if a != ai]
    scale = 1.0 if unit == 'grid' else step / 1e-10
    pts = [(c[0], c[1]) for c in charges if c[1][ai] == index]
    if not pts:
        print(f'no charges at {axis} = {index}'); return
    q = np.array([p[0] for p in pts])
    co = np.array([p[1] for p in pts], dtype=float) * scale

    if ax is None:
        _, ax = plt.subplots(figsize=(6, 5))
    ax.scatter(co[q > 0, other[0]], co[q > 0, other[1]], c='#d62728', s=40, label='+W')
    ax.scatter(co[q < 0, other[0]], co[q < 0, other[1]], c='#1f77b4', s=40, label='-W')
    names = ['x', 'y', 'z']
    label = 'grid points' if unit == 'grid' else 'angstrom'
    ax.set_xlabel(f'{names[other[0]]} ({label})'); ax.set_ylabel(f'{names[other[1]]} ({label})')
    ax.set_aspect('equal'); ax.legend()
    ax.set_title(f'slice {axis} = {index}')
    return ax


# Plot EVERY charge (no subsampling) so the checkerboard (chess) ordering is
# visible -- a random subsample would scramble the +/- alternation and look random.
plot_charges(inp)
# ... plus one plane, where the checkerboard alternation is clearest.
plot_slice(inp, axis='z', index=0)
plt.show()
````

````python
def expand_pairs_to_vec(pairs):
    """Lay charges out on a flat length-2**N_QUBITS vector indexed by bitstring."""
    out = np.zeros(2**N_QUBITS)
    for q, s in pairs:
        out[int(s, 2)] = q
    return out
````

````python
vec = expand_pairs_to_vec(convert_input_to_pairs(inp))
W_NORM = np.linalg.norm(vec)
vec /= W_NORM
````

## Quantum circuit

The circuit has three logical layers:

1. **State preparation** — `circ.initialize(vec)` loads the normalized
   amplitudes $\alpha_j$ into 
   $N_{\text{qubits}} = \log_2 M_x + \log_2 M_y + \log_2 M_z = 18$ qubits.
2. **3D QFT** — one QFT per coordinate register: `QFT(MX_QUBITS)` on `X_REG`,
   `QFT(MY_QUBITS)` on `Y_REG` and `QFT(MZ_QUBITS)` on `Z_REG`, 6 qubits each.
   Because the registers act on disjoint qubits, the three QFTs commute and
   together implement a 3D discrete Fourier transform of an
   $M_x \times M_y \times M_z$ array.
3. **Measurement** — `circ.measure_all()`, skipped when
   `EXACT_SIM = True`.

After the QFT, the probability of measuring bitstring
$\mathbf{k} = (k_x, k_y, k_z)$ is

$$ P(\mathbf{k}) \;=\; \frac{|\hat\rho(\mathbf{k})|^2}
   {\|\rho\|^2 \cdot M_x M_y M_z}, $$

so $|\hat\rho(\mathbf{k})|^2 = \|\rho\|^2 \cdot N_{\text{grid}} \cdot
P(\mathbf{k})$. 

````python
circ = QuantumCircuit(N_QUBITS)
circ.initialize(vec)
circ.append(QFT(MX_QUBITS), X_REG)
circ.append(QFT(MY_QUBITS), Y_REG)
circ.append(QFT(MZ_QUBITS), Z_REG)

if not EXACT_SIM:
    circ.measure_all()
````

### Running the simulation

Either draw `N_SHOTS` samples from the circuit (default) or compute the
exact statevector probabilities (`EXACT_SIM = True`). The output `cnts`
is a dict mapping `'xyz'` bitstrings to either integer counts (sampling)
or float probabilities (exact). Both branches feed into the same
downstream code via the `cnt3d` rescaling cell below.

````python
backend = AerSimulator()
tcirc = transpile(circ, backend, optimization_level=0)
print(f"Transpiled ops: {dict(tcirc.count_ops())}")

sampler = SamplerV2()

start = time.perf_counter()

if EXACT_SIM:
    result = Statevector(circ)
    cnts = result.probabilities_dict()
else:
    job = sampler.run([tcirc], shots=N_SHOTS)
    result = job.result()
    cnts = result[0].data['meas'].get_counts()

end = time.perf_counter()

print(f"Quantum emulator calculation time = {end - start:.6f} seconds")
````

## Processing results

The next three cells:

1. Reshape the measurement counts into an `(MX, MY, MZ)` reciprocal-space
   density `cnt3d`, applying the `W_NORM² / N_SHOTS` rescaling so that
   `cnt3d[k]` approximates $|\hat\rho(\mathbf{k})|^2 / N_{\text{grid}}$ in
   charge² units.
2. Compute the classical Fourier transform `cff` of the charge
   distribution as a reference.
3. Define `compute_e_l_matrix(σ, ff)`, the reciprocal-space Ewald
   kernel that turns either `|cff|²` or `cnt3d · N_grid` into the
   long-range energy $E^L$.

````python
# Reconstruct the reciprocal-space probability density on the (MX, MY, MZ) grid.
# In EXACT_SIM mode, `cnts` holds exact probabilities; otherwise empirical
# frequencies.
#
# For a bitstring 'xyz', int(bs, 2) is exactly the flat index
# (i << (MY_QUBITS + MZ_QUBITS)) | (j << MZ_QUBITS) | k, so the counts can be
# scattered straight into a flat array and reshaped -- which is why the register
# assignment above has to follow bit significance.
counts_flat = np.zeros(N_GRID, dtype=np.float64)
for bs, c in cnts.items():
    counts_flat[int(bs, 2)] = c
counts_array = counts_flat.reshape(SHAPE)

if EXACT_SIM:
    cnt3d = W_NORM**2 * counts_array
else:
    cnt3d = W_NORM**2 * counts_array / N_SHOTS
````

````python
start = time.perf_counter()
cff = fftn(expand_pairs_to_vec(convert_input_to_pairs(inp)).reshape(MX, MY, MZ))
end = time.perf_counter()

print(f"Classical FFT time = {end - start:.6f} seconds")
````

````python
def compute_e_l_matrix(sigma: float, ff: np.ndarray) -> np.ndarray:
    """Reciprocal-space Ewald kernel applied to a |FFT(rho)|^2 grid of shape SHAPE.

    Each axis carries its own fundamental wavevector 2*pi / L_axis.  The three
    coincide for the cubic box used here, but the per-axis form keeps the kernel
    valid for any M_x, M_y, M_z.
    """
    k_sq = np.fromfunction(
        lambda i, j, k: (2 * np.pi / (MX * step))**2 * i * i
                      + (2 * np.pi / (MY * step))**2 * j * j
                      + (2 * np.pi / (MZ * step))**2 * k * k,
        SHAPE,
        dtype=float,
    )
    coef = np.divide(
        np.exp(-sigma**2 * k_sq / 2), k_sq,
        where=(k_sq != 0),
        out=np.zeros_like(k_sq),
    )
    return coef * ff / (2 * VOL * eps0)


start = time.perf_counter()
e_l_classical = compute_e_l_matrix(SIGMA, np.abs(cff)**2).sum()
end = time.perf_counter()

print(f"E^L (classical) = {e_l_classical}")
print(f"Time: {end - start:.6f} sec")
````

## $E^S$ (real-space term) — split into *build list* + *calculate*

The real-space short-range contribution is

$$ E^S \;=\; \frac{1}{4\pi\varepsilon_0}
   \sum_{\substack{i<j \\ \mathbf{n}}}
   \frac{q_i q_j}{r_{ij,\mathbf{n}}}\,
   \mathrm{erfc}\!\left(\frac{r_{ij,\mathbf{n}}}{\sigma\sqrt{2}}\right), $$

where $r_{ij,\mathbf{n}}$ is the distance between charges $i$ and $j$ with $j$
taken from periodic image $\mathbf{n}$. Pairs with $r_{ij,\mathbf{n}} \ge$ `RCUT`
are dropped (the `erfc` factor makes the truncation error tiny).

The computation is split into the two steps below, mirroring how a
molecular-dynamics engine separates **neighbour-list construction** from the
**per-step force/energy evaluation**.

**Step 1 — build the neighbour list (`build_neighbor_lists`).** Depends on `RCUT`
(geometry) only. Using a 3D cell list of side `RCUT`:

1. Expand the original charges into themselves plus all periodic images out to $n$
   shells.
2. Bin all coordinates into cubic cells of side `RCUT`.
3. For each original charge, scan only its $3\times3\times3$ neighbourhood and store
   the **indices** of the in-cutoff partners (`idx_i`, `idx_j`) together with the
   coordinate/charge tables they point into. **No distances or charge products are
   cached** — only which pairs interact.

The list is stored in `NEIGHBORS` and stays valid for every $\sigma$, since $\sigma$
enters only the screening function.

**Step 2 — the actual $E^S$ calculation (`compute_e_s`).** Walks the neighbour list
and **recomputes each pair distance $r_{ij}$ from the coordinates** — exactly as an
MD engine recomputes distances at every step from the current positions — then
applies the $\sigma$-dependent `erfc` screening.

`Grid` (built in the next cell) is the un-normalized charge distribution as an
`(MX, MY, MZ)` array — reused by `compute_e_self`, `dipole_moment`, and
`compute_coulomb_grid`.

````python
Grid = expand_pairs_to_vec(convert_input_to_pairs(inp)).reshape(SHAPE)


def build_neighbor_lists():
    """Step 1 of E^S: build ONLY the neighbour list.

    For the chosen RCUT, find every in-cutoff pair (i, j) with j > i, where i is an
    original charge and j is any charge (original or periodic image).  Only the pair
    *indices* are stored -- NO distances and NO charge products are cached.  The
    distances are recomputed from the coordinates at energy-evaluation time (next
    cell), emulating a molecular-dynamics step where the positions move between
    neighbour-list rebuilds.  The list depends on RCUT (geometry) only -- not on SIGMA.
    """
    # --- Coordinates of the original charges and their periodic images ---
    xs, ys, zs = np.nonzero(Grid)
    q_vals = Grid[xs, ys, zs]
    original_coords = np.column_stack((xs, ys, zs)) * step

    shifts = np.arange(-n, n + 1)
    idx_x, idx_y, idx_z = np.meshgrid(shifts, shifts, shifts, indexing='ij')
    ox, oy, oz = idx_x.ravel(), idx_y.ravel(), idx_z.ravel()
    keep = ~((ox == 0) & (oy == 0) & (oz == 0))
    ox, oy, oz = ox[keep], oy[keep], oz[keep]

    # One box length per axis now (BOX = (MX, MY, MZ) * step), broadcast over
    # the three columns of the shift table.
    offsets = np.column_stack((ox, oy, oz)) * BOX
    image_coords = (original_coords[:, None, :] + offsets[None, :, :]).reshape(-1, 3)
    image_q = np.repeat(q_vals, len(offsets))

    final_coords = np.concatenate((original_coords, image_coords))
    final_q = np.concatenate((q_vals, image_q))

    # --- Build the 3D cell grid (cell side = RCUT) ---
    cell_indices = np.floor(final_coords / RCUT).astype(int)
    grid_map = defaultdict(list)
    for i, c_idx in enumerate(cell_indices):
        grid_map[tuple(c_idx)].append(i)

    max_r, max_c, max_h = cell_indices.max(axis=0) + 1
    grid_array = np.empty((max_r, max_c, max_h), dtype=object)
    grid_array[:] = np.frompyfunc(list, 0, 1)(np.empty((max_r, max_c, max_h), dtype=object))
    for (r, c, h), indices in grid_map.items():
        grid_array[r, c, h] = indices

    # --- For each original charge, record the in-cutoff partner INDICES (j > i) ---
    N_orig = len(q_vals)
    idx_i, idx_j = [], []
    for i in range(N_orig):
        ri = original_coords[i]
        cell_r, cell_c, cell_h = cell_indices[i]

        r_start, r_end = max(0, cell_r - 1), min(max_r, cell_r + 2)
        c_start, c_end = max(0, cell_c - 1), min(max_c, cell_c + 2)
        h_start, h_end = max(0, cell_h - 1), min(max_h, cell_h + 2)

        neighborhood = grid_array[r_start:r_end, c_start:c_end, h_start:h_end]
        neighbor_indices = []
        for cell_list in neighborhood.flatten():
            neighbor_indices.extend(cell_list)
        neighbor_indices = np.array(neighbor_indices)

        # Avoid double-counting between original charges.
        target_indices = neighbor_indices[neighbor_indices > i]
        if len(target_indices) == 0:
            continue

        # The cutoff membership test uses a distance, but ONLY the index pair is
        # stored -- the energy cell recomputes r_ij itself.
        d = np.linalg.norm(final_coords[target_indices] - ri, axis=1)
        sel = target_indices[d < RCUT]
        if len(sel) == 0:
            continue
        idx_i.append(np.full(len(sel), i))
        idx_j.append(sel)

    if idx_i:
        idx_i = np.concatenate(idx_i)
        idx_j = np.concatenate(idx_j)
    else:
        idx_i = np.array([], dtype=int)
        idx_j = np.array([], dtype=int)

    # Store ONLY the pair indices + the coordinate/charge tables they point into.
    return dict(idx_i=idx_i, idx_j=idx_j,
                coords=final_coords, charges=final_q,
                n_pairs=len(idx_i))


start = time.perf_counter()
NEIGHBORS = build_neighbor_lists()
end = time.perf_counter()

print(f"Neighbour list built for RCUT = {RCUT / 1e-10:.3f} A")
print(f"In-cutoff pairs: {NEIGHBORS['n_pairs']}")
print(f"Time: {end - start:.6f} sec")
````

````python
def compute_e_s() -> float:
    """Step 2 of E^S: real-space sum over the pre-built neighbour list.

    The pair distances are RECOMPUTED here from the current coordinates (rather than
    read from a cache), emulating a molecular-dynamics step: the neighbour list fixes
    *which* pairs interact, and every energy evaluation recomputes r_ij for those
    pairs from their positions.  Only the erfc screening depends on SIGMA.
    """
    idx_i   = NEIGHBORS['idx_i']
    idx_j   = NEIGHBORS['idx_j']
    coords  = NEIGHBORS['coords']
    charges = NEIGHBORS['charges']
    if len(idx_i) == 0:
        return 0.0

    # Recompute the neighbour-pair distances from the coordinates.
    r_ij  = np.linalg.norm(coords[idx_i] - coords[idx_j], axis=1)
    qi_qj = charges[idx_i] * charges[idx_j]

    coef = np.sqrt(2) * SIGMA
    return np.sum((qi_qj / r_ij) * erfc(r_ij / coef)) / (4 * np.pi * eps0)


start = time.perf_counter()
e_s = compute_e_s()
end = time.perf_counter()

print(f"E^S = {e_s}")
print(f"Time: {end - start:.6f} sec")
````

## $E^{\mathrm{self}}$ (self-interaction)

The Gaussian charge that screens each particle in the reciprocal-space
sum interacts with itself; subtracting that spurious self-energy gives

$$ E^{\mathrm{self}} \;=\; -\frac{1}{4\pi\varepsilon_0}\,
   \frac{1}{\sigma\sqrt{2\pi}} \sum_i q_i^2. $$

`compute_e_self(charges)` evaluates this as a single vectorized
`np.sum(charges**2)`. The argument can be any array of charges (here we
pass `Grid`); zeros do not contribute.

````python
def compute_e_self(charges: np.ndarray) -> float:
    """Self-interaction term E^self (vectorized sum of squared charges)."""
    sum_q2 = np.sum(charges**2)
    return -(sum_q2 / (SIGMA * np.sqrt(2) * np.sqrt(np.pi))) / (4 * np.pi * eps0)


start = time.perf_counter()
e_self = compute_e_self(Grid)
end = time.perf_counter()

print(f"E^self = {e_self}")
print(f"Time: {end - start:.6f} sec")
````

## $E^{\mathrm{dipole}}$ (surface correction)

When the simulation cell carries a net dipole moment, the reciprocal-space
sum (which assumes the surface at infinity is conducting, $\varepsilon_p
\to \infty$) needs a correction:

$$ E^{\mathrm{dipole}} \;=\; \frac{1}{4\pi\varepsilon_0}\,
   \frac{2\pi}{(1 + 2\varepsilon_p)\,V}\, |\mathbf{P}|^2,
   \qquad \mathbf{P} = \sum_i q_i\,\mathbf{r}_i, $$

with $V = M_x M_y M_z \Delta x^3$ the supercell volume.

For an overall neutral cell (the alternating-sign lattice used here)
$\mathbf{P} = \mathbf{0}$ and `E^dipole` is zero. The implementation
computes each component $P_\alpha$ as
`grid.sum(over the two axes orthogonal to α) @ coords_α`, which is the
vectorized form of $\sum_i q_i\,r_i^\alpha$ — one coordinate vector per axis, so
the expression stays valid for boxes whose axes differ in length.

````python
def dipole_moment(grid_data: np.ndarray) -> float:
    """Surface (dipole) correction term E^dipole."""
    grid_data = grid_data.astype(float)
    coords_x = (np.arange(MX) * step).astype(float)
    coords_y = (np.arange(MY) * step).astype(float)
    coords_z = (np.arange(MZ) * step).astype(float)

    # P_alpha = sum over charges of q_i * r_i_alpha. Reduce over the two axes
    # orthogonal to alpha, then take the dot product with that axis' coordinates.
    Px = np.dot(grid_data.sum(axis=(1, 2)), coords_x)
    Py = np.dot(grid_data.sum(axis=(0, 2)), coords_y)
    Pz = np.dot(grid_data.sum(axis=(0, 1)), coords_z)

    P_sq = Px**2 + Py**2 + Pz**2
    return (2 * np.pi / ((1 + 2 * epsp) * VOL * 4 * np.pi * eps0)) * P_sq


start = time.perf_counter()
e_dipole = dipole_moment(Grid)
end = time.perf_counter()

print(f"E^dipole = {e_dipole}")
print(f"Time: {end - start:.6f} sec")
````

## Direct Coulomb summation (ground truth)

A straightforward $\mathcal{O}(N^2)$ Coulomb sum, used as the reference
against which both Ewald variants are scored. Pairs are summed within
the unit cell ($i < j$, to avoid double-counting) and between the unit
cell and its periodic images out to $m$ shells:

$$ E_{\mathrm{Coulomb}} \;=\;
   \frac{1}{4\pi\varepsilon_0}\!
   \left[\sum_{i<j} \frac{q_i q_j}{r_{ij}}
   \;+\; \sum_{i,j,\,\mathbf{n}\neq 0} \frac{q_i q_j}{r_{ij,\mathbf{n}}}\right]. $$

This is feasible only because the test configurations are small. It
scales as $N^2 \cdot (2m+1)^3$ and becomes prohibitive for large
systems — which is the whole reason Ewald summation exists.

````python
def compute_coulomb_grid(grid: np.ndarray, m: int, shape, step: float,
                         eps0: float, chunk: int = CHUNK) -> float:
    """Direct O(N^2) Coulomb sum over the unit cell plus m image shells (ground truth).

    Both sums are accumulated in blocks of `chunk` rows, so the full (N x 26N)
    distance matrix is never held in memory at once; blocking changes the order of
    summation only, not the result.

    `shape` is (MX, MY, MZ): each image shift is scaled by its own axis length.
    """
    # 1. Coordinates and charges in the unit cell.
    xs, ys, zs = np.nonzero(grid)
    q_initial = grid[xs, ys, zs]
    coords_initial = np.stack([xs, ys, zs], axis=1).astype(float)
    N = len(q_initial)

    # 2. Periodic-image offsets, excluding the (0, 0, 0) cell.
    offsets = np.arange(-m, m + 1)
    mx, my, mz = np.meshgrid(offsets, offsets, offsets)
    shift_indices = np.stack([mx.ravel(), my.ravel(), mz.ravel()], axis=1)
    shift_indices = shift_indices[~np.all(shift_indices == 0, axis=1)]

    lengths = np.asarray(shape, dtype=float)          # (MX, MY, MZ) in grid units
    coords_periodic = np.vstack([coords_initial + s * lengths for s in shift_indices])
    q_periodic = np.concatenate([q_initial for _ in shift_indices])
    print(f"N = {N:,}   images = {len(q_periodic):,}   "
          f"pairs = {N * len(q_periodic):.3e}")

    # 3. Unit cell <-> periodic images (chunked over rows).
    summ = 0.0
    t0 = time.perf_counter()
    for a in range(0, N, chunk):
        b = min(a + chunk, N)
        d = cdist(coords_initial[a:b], coords_periodic, metric='euclidean') * step
        summ += np.sum((q_initial[a:b, np.newaxis] * q_periodic) / d)
        if (a // chunk) % 20 == 0:
            frac = b / N
            el = time.perf_counter() - t0
            print(f"  images {b:,}/{N:,} ({100 * frac:.1f}%)  "
                  f"{el:.0f}s elapsed, ~{el / max(frac, 1e-9) - el:.0f}s left")

    # 4. Self-interaction within the unit cell (i < j to avoid double counting),
    #    same chunking; the j > i mask is applied on global column indices.
    internal_summ = 0.0
    for a in range(0, N, chunk):
        b = min(a + chunk, N)
        d = cdist(coords_initial[a:b], coords_initial, metric='euclidean') * step
        qq = q_initial[a:b, np.newaxis] * q_initial
        mask = np.arange(N)[None, :] > np.arange(a, b)[:, None]
        internal_summ += np.sum(np.where(mask, qq / np.where(d == 0, 1.0, d), 0.0))

    return (summ + internal_summ) / (4 * np.pi * eps0)


start = time.perf_counter()
e_coulomb = compute_coulomb_grid(Grid, m, SHAPE, step, eps0)
end = time.perf_counter()

print(f"E_Coulomb = {e_coulomb}")
print(f"Time: {end - start:.6f} sec")
````

## Total PME energy (classical reference)

Recompute all four terms in one cell so that timings and intermediate
values are produced together, and sum them into

$$ E_{\mathrm{PME}} \;=\; E^L \;+\; E^S \;+\; E^{\mathrm{self}} \;+\; E^{\mathrm{dipole}}, $$

using the classical FFT for $E^L$. This is the all-classical baseline
that the quantum result will be compared against.

````python
start = time.perf_counter()

cff           = fftn(expand_pairs_to_vec(convert_input_to_pairs(inp)).reshape(SHAPE))
e_l_classical = compute_e_l_matrix(SIGMA, np.abs(cff)**2).sum()
e_s           = compute_e_s()
e_self        = compute_e_self(Grid)
e_dipole      = dipole_moment(Grid)

end = time.perf_counter()

E_PME = e_s + e_l_classical + e_self + e_dipole

print(f"PME calculation time = {end - start:.6f} seconds\n")
print(f"E_PME    = {E_PME}")
print(f"E^L      = {e_l_classical}")
print(f"E^S      = {e_s}")
print(f"E^self   = {e_self}")
print(f"E^dipole = {e_dipole}")
````

## Quantum vs classical $E^L$

The next cell repeats the $E^L$ calculation using the quantum-measured
density `cnt3d` instead of the classical FFT. Because the quantum state
is L2-normalized and probabilities sum to one, the recovered magnitude
differs from $|\hat\rho(\mathbf{k})|^2$ by exactly a factor of
$N_{\text{grid}} = M_x M_y M_z$, which is restored here:

```
e_l_quantum = compute_e_l_matrix(SIGMA, cnt3d).sum() * N_GRID
```

The cell also assembles `E_quantum = e_l_quantum + e_s + e_self + e_dipole`
— the same total as $E_{\mathrm{PME}}$ but with $E^L$ replaced — and
reports the gap $|E_{\mathrm{quantum}} - E_{\mathrm{PME}}|$, which is
entirely due to shot noise (and, if the FakeMontreal/Sherbrooke cells
were run, hardware noise).

````python
e_l_quantum = compute_e_l_matrix(SIGMA, cnt3d).sum() * N_GRID
E_quantum   = e_l_quantum + e_s + e_self + e_dipole

print(f"E^L (quantum)   = {e_l_quantum}")
print(f"E^L (classical) = {e_l_classical}")
print(f"|q - c|         = {abs(e_l_quantum - e_l_classical)}\n")
print(f"E_PME (quantum) = {E_quantum}")
print(f"Quantum error   = {np.abs(E_quantum - E_PME)}")
````

## Final comparison

Compute the absolute deltas $E_{\mathrm{PME}} - E_{\mathrm{Coulomb}}$ and
$E_{\mathrm{quantum}} - E_{\mathrm{Coulomb}}$, and the dimensionless
relative errors

$$ \mathrm{rms}_E \;=\; \sqrt{
   \frac{\bigl(E_{\mathrm{Coulomb}} - E_{\mathrm{candidate}}\bigr)^2}
        {E_{\mathrm{Coulomb}}^2}
   }. $$

`rms_E (classical)` is the floor (Ewald truncation + FFT discretization
error). `rms_E (quantum)` adds the shot-noise contribution from the
measurement-based reconstruction of $|\hat\rho(\mathbf{k})|^2$.

````python
delta_E   = E_PME     - e_coulomb
delta_E_q = E_quantum - e_coulomb

print(f"E_PME      - E_Coulomb = {delta_E}")
print(f"E_quantum  - E_Coulomb = {delta_E_q}\n")
print(f"rms_E (classical) = {np.sqrt((e_coulomb - E_PME)**2 / e_coulomb**2)}")
print(f"rms_E (quantum)   = {np.sqrt((e_coulomb - E_quantum)**2 / e_coulomb**2)}")
````

## Quantum circuit complexity

The depth of the circuit is the number of "layers". This definition depends on the set of allowed gates, the number of available qubits (we can use free qubits as ancillae to simplify certain gates), and the QPU topology (e.g. if we allow any state preparation and QFT gates, we get the depth = 3). Therefore, we have to fix the quantum chip, which also fixes the allowed gates and the topology.

Here, we compile the original circuit for the "Sherbrooke" QPU (it has 127 qubits) from IBM.
